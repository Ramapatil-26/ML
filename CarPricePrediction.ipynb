```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
from sklearn.linear_model import LinearRegression
```

```python
#Load dataset
df=pd.read_csv("car_price_prediction.csv")
print(df.head())
```

```python
plt.scatter(df["Year"],df["Price"])
```

```python
plt.scatter(df["Engine Size"],df["Price"])
```

```python
plt.scatter(df["Mileage"],df["Price"])
```

```python
df.corr()
```

```python
sns.pairplot(df)
```

```python
X = df[['Year', 'Engine Size', 'Mileage',]]
y = df['Price']
```

```python
X_train,X_test,y_train,y_test=train_test_split(X,y,train_size=0.8,test_size=42)
```

```python
model=LinearRegression()
model.fit(X_train,y_train)
```

```python
print("Intercept",model.intercept_)
```

```python
print("Coefficent",model.coef_)
```

```python
y_pred=model.predict(X_test)
print("Predicted Value",y_train)
print("Actual Value",y_test.values)
```

```python
mae=mean_absolute_error(y_test,y_pred)
mse=mean_squared_error(y_test,y_pred)
rmse=np.sqrt(mse)
r2=r2_score(y_test,y_pred)

print("MAE: ", mae)
print("MSE: ", mse)
print("RMSE: ",rmse)
print("R2: ", r2)
```

```python
car=[[2025, 0.8, 50000]]
predict=model.predict(car)
print("Predicted Price: ",predict[0])
```

```python
year=int(input("Enter the Year: "))
mileage=int(input("Enter the Mileage value: "))
engine_size=int(input("Enter the Engine size: "))

predicted_price=model.predict([[year,mileage,engine_size]])
print("Predicted Price: ",predicted_price[0])
```
