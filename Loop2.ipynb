```python
#looping through list
fruits = ["Cherry", "Banana", "Orange"]
for x in fruits:
    print(x)
```

```python
#looping through a string
for i in "Orange":
    print(i)
```

```python
#break looping
fruits = ["Cherry", "Banana", "Orange"]
for x in fruits:
    #print(x)
    if x == "Banana":
        break
    print(x)
```

```python
#continue looping
fruits = ["Cherry", "Banana", "Orange"]
for x in fruits:
    if x == "Banana":
        continue
    print(x)
```

```python
#nested loop
adj = ["Red","Big","Tasty"]
fruits = ["Apple","Banana","Cherry"]
for x in adj:
    for y in fruits:
        print(x,y)
```
