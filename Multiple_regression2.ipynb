```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
from sklearn.linear_model import LinearRegression
```

Create Dataset

```python
data = {
    "Size" : [700,900,1050,1050,1200,1300,1300,1500,1700,1700,2000,2200],
    "BHK" :  [1,1,1,2,2,2,3,3,3,4,4,4 ],
    "Price" :[1400000,1700000,1950000,2025000,2450000,2700000,2800000,3200000,3500000,3900000,4200000,4400000]
}
```

```python
print(type(data))
```

```python
df=pd.DataFrame(data)
```

```python
print(df)
```

```python
plt.scatter(df["Size"],df["BHK"])
```

```python
plt.scatter(df["Size"],df["Price"])
```

```python
plt.scatter(df["Price"],df["BHK"])
```

```python
df.corr()
```

Seaborn Visualization

```python
sns.pairplot(df)
```

Finalized Dependant and independent features

```python
X = df[["Size","BHK"]]
y = df[["Price"]]
```

Split the dataset for training and testing

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.8,random_state=42)
```

Train the Model

```python
model = LinearRegression()
model.fit(X_train,y_train)
```

view Model Parameters

```python
print("intercept:",model.intercept_)
```

```python
print("co efficent:",model.coef_)
```

Make the Predictions

```python
y_pred = model.predict(X_test)
print("predict value",y_train)
print("actual value",y_test.values)
```

Evaluate model

```python
mse = mean_squared_error(y_test,y_pred)
```

```python
r2 = r2_score(y_test,y_pred)
```

predict price for new flate

```python
a =[[1700,3]]
p = model.predict(a)
print("predicted price = ",p)
```
